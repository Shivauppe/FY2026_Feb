```python
import pandas as pd

from scipy import stats
```

```python
# 11.  What can be concluded from the output if the printed p-value is 0.0002?  


batch_A = [82,85,90,88,86]
batch_B = [78,76,80,79,77]
batch_c = [91,89,92,94,90]

f_stat,p_value = stats.f_oneway(batch_A,batch_B,batch_c)
print(p_value)

# p-value is less than 0.05, 
# we can reject the null hypothesis a
# and conclude that there is a significant difference in the mean scores of the three batches.
```
